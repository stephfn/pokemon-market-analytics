## API Data Integration

### Objective

Pokémon attribute data were retrieved directly from the PokéAPI in JSON format and transformed into a structured analytical dataset. Several cleaning and standardization steps were performed to improve readability, consistency, and compatibility with the relational database used later in the project.

## API Data Source

Pokémon data was pulled directly from the PokéAPI JSON source using the `requests` library. The API response was converted into a pandas DataFrame so the data could be cleaned, formatted, and validated without exporting it to a CSV file.

```python
import requests
import pandas as pd

# Pull Pokémon data directly from the API
url = "https://pokeapi.co/api/v2/pokemon?limit=150"

response = requests.get(url)

# Check that the API request worked
print("Status code:", response.status_code)

# Convert JSON response to Python dictionary
data = response.json()

# Extract Pokémon list from JSON response
pokemon_list = data["results"]

# Convert API results into a DataFrame
df_api = pd.DataFrame(pokemon_list)

print("Dataset shape:", df_api.shape)

df_api.head()
```

## Normalize API JSON Data

The API response was retrieved in JSON format and converted into a pandas DataFrame. This transformed the nested JSON response into a structured format that is easier to inspect and clean.

```python
# Inspect normalized API dataset
df_api.info()
```

## Standardize Column Headers

```python
df_api.columns = (
    df_api.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df_api.columns)
```

## Standardize Pokémon Name Formatting

Pokémon names were converted to title case to improve readability and create consistent text formatting.

```python
df_api["name"] = df_api["name"].str.title()

df_api.head()
```

## Extract Pokémon ID from API URL

The Pokémon ID was extracted from the API URL using regular expressions to create a usable numeric identifier for each Pokémon record.

```python
df_api["pokemon_id"] = df_api["url"].str.extract(r"/pokemon/(\d+)/").astype(int)

df_api.head()
```

## Extract Additional Pokémon Attributes

Additional Pokémon attributes were retrieved from each Pokémon-specific API endpoint to create a richer dataset for cleaning and analysis.

```python
detailed_pokemon = []

for url in df_api["url"][:25]:
    pokemon_response = requests.get(url)
    pokemon_data = pokemon_response.json()

    detailed_pokemon.append({
        "name": pokemon_data["name"].title(),
        "height": pokemon_data["height"],
        "weight": pokemon_data["weight"],
        "base_experience": pokemon_data["base_experience"],
        "primary_type": pokemon_data["types"][0]["type"]["name"].title()
    })

detailed_df = pd.DataFrame(detailed_pokemon)

detailed_df.head()
```

## Convert Height Units

Pokémon height values were converted from decimeters to meters to make the measurements easier to interpret.

```python
detailed_df["height_m"] = detailed_df["height"] / 10

# Rename raw API measurement columns
detailed_df = detailed_df.rename(columns={
    "height": "height_dm",
    "weight": "weight_hg"
})

detailed_df.head()
```

## Convert Weight Units

Pokémon weight values were converted from hectograms to kilograms to improve readability and standardize units for analysis.

```python
# Convert weight from hectograms to kilograms
detailed_df["weight_kg"] = detailed_df["weight_hg"] / 10

detailed_df.head()
```

The original API weight values were preserved as `weight_hg`, while the converted `weight_kg` column provides a more interpretable measurement for analysis.

## Check for Missing Values

Missing values were checked to evaluate whether the API response contained incomplete records.

```python
# Check for missing values in the detailed API dataset
missing_values = detailed_df.isnull().sum()

print(missing_values)

# Display rows containing missing values (if any exist)
missing_rows = detailed_df[detailed_df.isnull().any(axis=1)]

print(missing_rows)
```

No missing values were identified in the API dataset. This confirmed that the extracted records were complete for the selected fields.

## Check for Duplicate Records

Duplicate records were checked to ensure that each Pokémon entry appeared only once in the API dataset.

```python
# Check for duplicate records in the detailed API dataset
print("Duplicate rows:", detailed_df.duplicated().sum())

# Check for duplicate Pokémon names
print("Duplicate Pokémon names:", detailed_df["name"].duplicated().sum())
```

No duplicate records or duplicate Pokémon names were identified in the detailed API dataset. This confirmed that each extracted Pokémon entry represented a unique record.

```python
# Create working copy of cleaned Pokémon statistics dataset
website_df = df.copy()

# Create working copy of cleaned PokéAPI dataset
api_df = detailed_df.copy()

# Load Pokémon card marketplace dataset
cards_df = pd.read_csv("pokemon_cards_ultimate_2026.csv")

print("Website dataset shape:", website_df.shape)
print("API dataset shape:", api_df.shape)
print("Card dataset shape:", cards_df.shape)
```

Column names were standardized to lowercase formatting with underscores to improve readability and consistency during SQL integration.

## Standardize Pokémon Name Formatting

Pokémon names were converted to title case to ensure consistency between datasets prior to SQL joins.

## Database Construction

After cleaning and standardizing each dataset, the data were loaded into SQLite as independent relational tables. SQLite was selected because it provides a lightweight relational database system that supports SQL-based data integration and analysis.

Three tables were created:

- pokemon_website
- pokemon_api
- pokemon_cards

These tables were subsequently joined using SQL to create a consolidated analytical dataset.

## Create SQLite Database Connection

A SQLite database connection was created to store each cleaned dataset as an individual SQL table.

```python
conn = sqlite3.connect("pokemon_project.db")
```

```python
website_df = website_df.rename(columns={"type": "pokemon_type"})
```

## Store Cleaned Datasets as SQLite Tables

Each cleaned dataset was stored as an individual table within the SQLite database.

```python
website_df.to_sql(
    "pokemon_website",
    conn,
    if_exists="replace",
    index=False
)

api_df.to_sql(
    "pokemon_api",
    conn,
    if_exists="replace",
    index=False
)

cards_clean.to_sql(
    "pokemon_cards",
    conn,
    if_exists="replace",
    index=False
)
```

```python
# Verify tables were created

cursor = conn.cursor()

cursor.execute("""
SELECT name
FROM sqlite_master
WHERE type='table';
""")

print(cursor.fetchall())
```

The output confirms that all three cleaned Pokémon datasets were successfully stored as individual tables within the SQLite database and are available for SQL-based integration and analysis.

## SQL Data Integration

The three Pokémon datasets were integrated using SQL INNER JOIN operations on standardized Pokémon name fields. This approach preserved relationships between game statistics, API-derived attributes, and trading card marketplace information while enabling analysis across multiple independent data sources.

The resulting merged dataset combined battle statistics, Pokémon attributes, rarity classifications, card conditions, and marketplace pricing information into a unified analytical view that served as the foundation for all subsequent visualizations and exploratory analysis.

### Join Strategy

An INNER JOIN was used to combine the three datasets based on Pokémon name fields. This ensured that only Pokémon present across all sources were included in the final analytical dataset. The join preserved relationships between game statistics, API attributes, and card marketplace information while maintaining data consistency.

### SQL Query

```python
query = """
SELECT
    w.name,
    w.pokemon_type,
    w.total,
    a.height_m,
    a.weight_kg,
    a.base_experience,
    c.rarity_class,
    c.price_usd,
    c.condition_std
FROM pokemon_website w
INNER JOIN pokemon_api a
    ON w.name = a.name
INNER JOIN pokemon_cards c
    ON w.name = c.pokemon_name;
"""

merged_df = pd.read_sql_query(query, conn)

merged_df.head()
```

### Merged Dataset Preview

The merged analytical dataset contains variables originating from all three source systems, demonstrating successful relational integration within SQLite. The resulting dataset enables direct comparison of Pokémon characteristics, rarity classifications, and marketplace pricing behavior.

### Interpretation

The SQL joins successfully combined Pokémon statistics, API attributes, and trading card pricing information into a consolidated dataset. The merged dataset allows exploratory analysis between Pokémon characteristics, rarity classifications, and marketplace pricing behavior.

# Average Card Price by Pokémon Type

```python
avg_price_type = (
    merged_df.groupby("pokemon_type")["price_usd"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

avg_price_type.plot(kind="bar")

plt.title("Average Pokémon Card Price by Pokémon Type")
plt.xlabel("Pokémon Type")
plt.ylabel("Average Card Price (USD)")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "avg_price_by_type.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Key Findings

- Fire and Electric type Pokémon exhibited some of the highest average card prices.
- Normal and Bug type Pokémon generally showed lower average market values.
- These differences suggest that collector demand and rarity may influence card pricing beyond battle performance alone.

# Total Stats vs Card Price

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

# Scatterplot using merged SQL dataset
plt.scatter(
    merged_df["total"],
    merged_df["price_usd"],
    alpha=0.7
)

# Regression line
m, b = np.polyfit(
    merged_df["total"],
    merged_df["price_usd"],
    1
)

plt.plot(
    merged_df["total"],
    m * merged_df["total"] + b
)

plt.xlabel("Total Pokémon Stats")
plt.ylabel("Card Price (USD)")
plt.title("Total Pokémon Stats vs Card Price")

plt.tight_layout()

plt.savefig(
    "stats_vs_price.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Key Findings

- A weak positive relationship was observed between total Pokémon battle statistics and card price.
- While stronger Pokémon occasionally corresponded to higher card values, considerable variability remained.
- This suggests that rarity, condition, grading status, and collector demand likely have a greater influence on market pricing than battle statistics alone.

# Average Price by Rarity Class
